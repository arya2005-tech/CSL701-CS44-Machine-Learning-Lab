```python
# Summary statistics for Age and Fare

print("AGE STATISTICS")
print("Mean Age   :", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\nFARE STATISTICS")
print("Mean Fare   :", df["Fare"].mean())
print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())
```

Task 8 — Identify Missing Values|

```python
# Task 8: Identify missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Display only columns having missing values

missing_values[missing_values > 0]
```

#Phase 4: Data Manipulation & Cleaning
Task 9 — Fill Missing Age Values with Median

```python

age_median = df["Age"].median()

print("Median Age:", age_median)

# Fill missing Age values
df["Age"] = df["Age"].fillna(age_median)

print("\nMissing Age values after cleaning:",
      df["Age"].isnull().sum())
```

Task 10 — Filter Surviving Passengers

```python
# Task 10: Filter passengers who survived

survived_passengers = df[df["Survived"] == 1]

print("Number of surviving passengers:",
      len(survived_passengers))

survived_passengers.head(10)
```

```python
# Count survived and non-survived passengers

print(df["Survived"].value_counts())
```

Task 11 — Convert Fare Column to NumPy Array

```python

fare_array = df["Fare"].to_numpy()

print("Fare NumPy array:")
print(fare_array)

print("\nType:", type(fare_array))
print("Number of fare values:", len(fare_array))
```

Task 12 — Calculate Maximum and Average Fare

```python

maximum_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", maximum_fare)
print("Average Fare:", average_fare)
```

```python
print(f"Maximum Fare: {maximum_fare:.2f}")
print(f"Average Fare: {average_fare:.2f}")
```

#Phase 5: Data Visualization
Task 13 — Histogram of Age

```python

plt.figure(figsize=(10, 6))

plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

Task 14 — Bar Chart of Survival Count

```python

survival_counts = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    ["Did Not Survive", "Survived"],
    survival_counts.values,
    color=["blue", "skyblue"],
    edgecolor="black"
)

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

Task 15 — Scatter Graph of Age vs Fare

```python


plt.figure(figsize=(10, 6))

plt.scatter(
    df["Age"],
    df["Fare"],
    alpha=0.6,
    color="purple"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```
